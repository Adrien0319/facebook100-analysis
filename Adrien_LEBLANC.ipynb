```python
import matplotlib.pyplot as plt

def analyze_graph(G, name):
    degrees = [d for n, d in G.degree()]

    # Histogramme
    plt.hist(degrees, bins=50)
    plt.title(f"Degree Distribution - {name}")
    plt.xlabel("Degree")
    plt.ylabel("Frequency")
    plt.show()

    # Clustering
    global_clustering = nx.transitivity(G)
    local_clustering = nx.average_clustering(G)

    # Densité
    density = nx.density(G)

    print(f"\n--- {name} ---")
    print("Global clustering:", global_clustering)
    print("Local clustering:", local_clustering)
    print("Density:", density)


analyze_graph(G_caltech, "Caltech")
analyze_graph(G_mit, "MIT")
analyze_graph(G_jh, "Johns Hopkins")
```

```python
from abc import ABC, abstractmethod
import math
import networkx as nx

class LinkPrediction(ABC):
    def __init__(self, graph):
        self.graph = graph
        self.N = len(graph)

    def neighbors(self, v):
        return list(self.graph.neighbors(v))

    @abstractmethod
    def score(self, u, v):
        pass


class CommonNeighbors(LinkPrediction):
    def score(self, u, v):
        return len(set(self.neighbors(u)) & set(self.neighbors(v)))


class Jaccard(LinkPrediction):
    def score(self, u, v):
        Nu = set(self.neighbors(u))
        Nv = set(self.neighbors(v))
        union = Nu | Nv
        if len(union) == 0:
            return 0
        return len(Nu & Nv) / len(union)


class AdamicAdar(LinkPrediction):
    def score(self, u, v):
        common = set(self.neighbors(u)) & set(self.neighbors(v))
        s = 0
        for w in common:
            deg = self.graph.degree(w)
            if deg > 1:
                s += 1 / math.log(deg)
        return s
```

```python
import random
import pandas as pd

def normalize_edge(e):
    u, v = e
    return tuple(sorted((u, v)))

def remove_edges(G, fraction, seed=42):
    random.seed(seed)
    G_train = G.copy()
    edges = list(G_train.edges())
    n_remove = int(len(edges) * fraction)
    removed = random.sample(edges, n_remove)
    G_train.remove_edges_from(removed)
    removed_set = set(normalize_edge(e) for e in removed)
    return G_train, removed_set

def evaluate_predictor(G_train, removed_edges, predictor_class, ks):
    predictor = predictor_class(G_train)
    nodes = list(G_train.nodes())
    scores = []

    for i in range(len(nodes)):
        for j in range(i + 1, len(nodes)):
            u, v = nodes[i], nodes[j]
            if not G_train.has_edge(u, v):
                s = predictor.score(u, v)
                if s > 0:
                    scores.append((s, normalize_edge((u, v))))

    scores.sort(reverse=True, key=lambda x: x[0])

    results = []
    for k in ks:
        top_edges = set(edge for _, edge in scores[:k])
        tp = len(top_edges & removed_edges)

        precision = tp / k
        recall = tp / len(removed_edges)
        topk_rate = precision

        results.append({
            "k": k,
            "TP": tp,
            "precision": precision,
            "recall": recall,
            "top@k": topk_rate
        })

    return results

def run_link_prediction_experiment(G, graph_name):
    fractions = [0.05, 0.10, 0.15, 0.20]
    ks = [50, 100, 200, 300, 400]

    methods = {
        "Common Neighbors": CommonNeighbors,
        "Jaccard": Jaccard,
        "Adamic-Adar": AdamicAdar
    }

    all_results = []

    for f in fractions:
        G_train, removed = remove_edges(G, f)

        for method_name, method_class in methods.items():
            res = evaluate_predictor(G_train, removed, method_class, ks)

            for row in res:
                row["graph"] = graph_name
                row["fraction_removed"] = f
                row["method"] = method_name
                all_results.append(row)

    return pd.DataFrame(all_results)
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch_geometric.nn import GCNConv
from torch_geometric.utils import from_networkx

# Convert graph
data = from_networkx(G_caltech)
data.x = torch.eye(data.num_nodes)  # features simples (identity)

# Créer edges (train)
edge_index = data.edge_index

class GCN(nn.Module):
    def __init__(self, in_channels, hidden_channels):
        super().__init__()
        self.conv1 = GCNConv(in_channels, hidden_channels)
        self.conv2 = GCNConv(hidden_channels, hidden_channels)

    def forward(self, x, edge_index):
        x = self.conv1(x, edge_index)
        x = F.relu(x)
        x = self.conv2(x, edge_index)
        return x

model = GCN(data.num_nodes, 16)
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

# Training simple
for epoch in range(50):
    model.train()
    optimizer.zero_grad()
    out = model(data.x, edge_index)

    # reconstruction loss
    loss = (out @ out.T).mean()
    loss.backward()
    optimizer.step()

print("Training done")
```

```python
import random
import numpy as np

def mask_labels(df, column, fraction):
    df_copy = df.copy()
    indices = df_copy.index.tolist()

    num_remove = int(len(indices) * fraction)
    removed_idx = random.sample(indices, num_remove)

    true_values = df_copy.loc[removed_idx, column].copy()
    df_copy.loc[removed_idx, column] = -1  # label manquant

    return df_copy, removed_idx, true_values
```

```python
def label_propagation(G, df, column):
    labels = df.set_index("node_id")[column].to_dict()
    labels = {str(k): v for k, v in labels.items()}

    new_labels = labels.copy()

    for node in G.nodes():
        if labels.get(node) == -1:
            neighbors = list(G.neighbors(node))
            neighbor_labels = [labels[str(n)] for n in neighbors if labels.get(str(n)) != -1]

            if len(neighbor_labels) > 0:
                new_labels[node] = max(set(neighbor_labels), key=neighbor_labels.count)

    return new_labels
```

```python
def compute_accuracy(pred_labels, true_values, removed_idx):
    correct = 0

    for idx in removed_idx:
        if pred_labels.get(str(idx)) == true_values.loc[idx]:
            correct += 1

    return correct / len(removed_idx)
```

```python
fractions = [0.1, 0.2, 0.3]

for col in ["dorm", "major_index", "gender"]:
    print(f"\n--- {col} ---")

    for f in fractions:
        df_masked, removed_idx, true_vals = mask_labels(df_caltech, col, f)

        pred = label_propagation(G_caltech, df_masked, col)
        acc = compute_accuracy(pred, true_vals, removed_idx)

        print(f"fraction={f}, accuracy={acc:.3f}")
```

```python
def compute_mae(pred_labels, true_values, removed_idx):
    errors = []

    for idx in removed_idx:
        pred = pred_labels.get(str(idx))
        true = true_values.loc[idx]
        if pred is not None:
            errors.append(abs(pred - true))

    return sum(errors) / len(errors)
```

```python
fractions = [0.1, 0.2, 0.3]

for col in ["dorm", "major_index", "gender"]:
    print(f"\n--- {col} ---")

    for f in fractions:
        df_masked, removed_idx, true_vals = mask_labels(df_caltech, col, f)
        pred = label_propagation(G_caltech, df_masked, col)

        acc = compute_accuracy(pred, true_vals, removed_idx)
        mae = compute_mae(pred, true_vals, removed_idx)

        print(f"fraction={f}, accuracy={acc:.3f}, MAE={mae:.3f}")
```
